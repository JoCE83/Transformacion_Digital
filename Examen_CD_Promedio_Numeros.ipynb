```python

## Archivo 2: promedio.py

def calcular_promedio(numeros: list[float]) -> float:
    """
    Calcula el promedio de una lista de números.

    Parámetros:
        numeros (list[float]): Lista de números enteros o decimales.

    Retorna:
        float: Promedio de los números ingresados.

    Excepciones:
        ValueError: Si la lista está vacía.
    """
    if not numeros:
        raise ValueError("La lista no puede estar vacía.")

    promedio = sum(numeros) / len(numeros)
    return promedio


if __name__ == "__main__":
    calificaciones = [7.5, 9.3, 4.67, 9.0]
    resultado = calcular_promedio(calificaciones)
    print(f"El promedio de sus calificaciones es: {resultado}")
```
